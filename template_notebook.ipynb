## 4. ⚡ Transición a PyTorch Moderno

Una vez que comprendemos la mecánica interna, vemos cómo los frameworks industriales encapsulan este patrón de forma vectorizada, paralela y compatible con GPU.

```python
# Implementación equivalente idiomática en PyTorch
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Usando dispositivo:", device)

# Ejemplo de módulo PyTorch
class ModuloPyTorch(torch.nn.Module):
    def __init__(self):
        super().__init__()
        
    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return x

t_model = ModuloPyTorch().to(device)
t_out = t_model(torch.tensor([1.0, 2.0, 3.0], device=device))
print("Salida PyTorch:", t_out)
```
